#### Historical Data Ingestion

```python
catalog = "ws_restaurantanalyticsplatform_7405614613479137"

historical_orders_df = (
    spark.read
         .option("header", "true")
         .option("inferSchema", "true")
         .csv("/Volumes/ws_restaurantanalyticsplatform_7405614613479137/datafiles/data/historical_orders.csv")
)

# Rename columns to match Bronze schema
historical_orders_df = (
    historical_orders_df
        .withColumnRenamed("timestamp", "order_timestamp")
        .drop("created_at")
)

display(historical_orders_df)

(
    historical_orders_df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(f"{catalog}.01_bronze.historical_orders")
)

print("historical_orders loaded successfully")
```
