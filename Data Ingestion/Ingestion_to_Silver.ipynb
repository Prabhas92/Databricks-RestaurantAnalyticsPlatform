#### Customer Data Ingestion

```python
catalog = "ws_restaurantanalyticsplatform_7405614613479137"

customers_df = (
    spark.read
         .option("header", "true")
         .option("inferSchema", "true")
         .csv("/Volumes/ws_restaurantanalyticsplatform_7405614613479137/datafiles/data/customers.csv")
)

display(customers_df)

(
    customers_df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(f"{catalog}.02_silver.dim_customers")
)

print("customers loaded successfully")
```
